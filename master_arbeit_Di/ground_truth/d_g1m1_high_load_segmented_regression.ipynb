```python
"""
d_g1m1_high_load_segmented_regression.py

Segmented regression helper for G1M1 high-load GT where the period
2022-05-25 to 2023-01-28 is excluded due to data-quality issues.
"""

from pathlib import Path
from master_arbeit_Di.ground_truth.d_gt_raw_segmented import (
    SegmentedRegressionConfig,
    read_daily_mean_csv,
    process_segmented_gt_raw,
)
```

```python
CSV_FILE = r"master_arbeit_Di\ground_truth\output_backup\gt_raw_processed\G1M1_new__gt_diagram__g1m1_high_load__iref_1p48__tref_57__ohref_100__daily_mean.csv"
OUTPUT_DIR = r"master_arbeit_Di\ground_truth\output_backup\gt_raw_processed"
DATASET_NAME = "G1M1_new__gt_diagram__g1m1_high_load__iref_1p48__tref_57__ohref_100__daily_mean"
```

```python


# Segmented configuration: skip 2022-05-25 to 2023-01-28
SEGMENTED_CONFIG = SegmentedRegressionConfig(
    gap_start_date="2022-05-25",
    gap_end_date="2023-01-28",
    exclude_gap_start=True,   # exclude the start date itself
    exclude_gap_end=True,     # exclude the end date itself
)
```

```python

print(f"Processing: {CSV_FILE}")
print(f"Gap excluded: {SEGMENTED_CONFIG.gap_start_date} to {SEGMENTED_CONFIG.gap_end_date}")

if not Path(CSV_FILE).exists():
    print(f"ERROR: File not found: {CSV_FILE}")
    exit(1)

full_coverage_df, metrics = process_segmented_gt_raw(
    daily_mean_csv=CSV_FILE,
    config=SEGMENTED_CONFIG,
    output_dir=OUTPUT_DIR,
    dataset_name=DATASET_NAME,
    save_html=True,
)

# Display metrics
print("\n" + "="*70)
print("SEGMENTED REGRESSION SUMMARY")
print("="*70)

print(f"\nSegment 1 (before {SEGMENTED_CONFIG.gap_start_date}):")
print(f"  Points: {metrics['segment1']['n_points']}")
print(f"  Slope: {metrics['segment1']['slope_uv_per_h']:.3f} uV/h")
print(f"  R2: {metrics['segment1']['r2']:.4f}")
print(f"  Range: {metrics['segment1']['from']} to {metrics['segment1']['to']}")

print(f"\nExcluded Gap: {metrics['gap_excluded']['start']} to {metrics['gap_excluded']['end']}")

print(f"\nSegment 2 (from {SEGMENTED_CONFIG.gap_end_date} onwards):")
print(f"  Points: {metrics['segment2']['n_points']}")
print(f"  Slope: {metrics['segment2']['slope_uv_per_h']:.3f} uV/h")
print(f"  R2: {metrics['segment2']['r2']:.4f}")
print(f"  Range: {metrics['segment2']['from']} to {metrics['segment2']['to']}")

print(f"\nFull-coverage GT points generated: {metrics['full_coverage_points']}")
print("\nProcessing complete!")
```
